🔹 Selecting Multiple Columns

```python
df = pd.read_csv("rainfall.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# First 5 rows
print("First 5 rows:")
print(df.head())

# Last 5 rows
print("\nLast 5 rows:")
print(df.tail())
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nColumns having missing values:")
print(df.columns[df.isnull().any()])
```

4. Fill missing values

```python
# Numerical columns
num_cols = df.select_dtypes(include="number").columns

# Fill numerical missing values with mean
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

# Categorical columns
cat_cols = df.select_dtypes(include="object").columns

# Fill categorical missing values with mode
for col in cat_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

5. Remove duplicate rows

```python
df = df.drop_duplicates()

print("\nDuplicates removed.")
print("Rows after removing duplicates:", len(df))
```

## 🔹 Intermediate Questions

1. Select multiple columns and:
   - Fill missing values together

2. Apply transformation:
   - Increase salary/marks by 10%

3. Fix inconsistent text:
   - "Mumbai", "mumbai", "MUMBAI" → one format

4. Create new column:
   - Example: "Pass/Fail" based on marks

5. Filter data:
   - Students with marks > 80
   - Rainfall > 10

6. Drop unnecessary columns:
   - Based on relevance

7. Apply One-Hot Encoding:
   - On categorical columns

8. Sort dataset:
   - Based on date or marks

```python
import pandas as pd

# Load CSV file
df = pd.read_csv("rainfall.csv")

# Clean column names
df.columns = df.columns.str.strip()

print("Original Data:")
print(df.head())


# ============================================
# 1. Select multiple columns and fill missing values
# ============================================

cols = ["Year", "average_rain_fall_mm_per_year"]

# Convert to numeric
for col in cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Fill missing values with mean
df[cols] = df[cols].fillna(df[cols].mean())

print("\nAfter filling missing values:")
print(df.isnull().sum())


# ============================================
# 2. Apply transformation - Increase rainfall by 10%
# ============================================

df["average_rain_fall_mm_per_year"] = (
    df["average_rain_fall_mm_per_year"] * 1.10
)

print("\nAfter increasing rainfall by 10%:")
print(df.head())


# ============================================
# 3. Fix inconsistent text
# ============================================

# Convert Area names to lowercase and remove spaces
df["Area"] = df["Area"].astype(str).str.strip().str.lower()

print("\nUnique Area names:")
print(df["Area"].unique())


# ============================================
# 4. Create a new column
# ============================================

# Create rainfall category
df["Rainfall_Category"] = df["average_rain_fall_mm_per_year"].apply(
    lambda x: "High" if x > 1000 else "Low"
)

print("\nRainfall Category:")
print(df[["Area", "average_rain_fall_mm_per_year", "Rainfall_Category"]].head())


# ============================================
# 5. Filter data - Rainfall > 10
# ============================================

high_rainfall = df[
    df["average_rain_fall_mm_per_year"] > 10
]

print("\nAreas with rainfall greater than 10:")
print(high_rainfall)


# ============================================
# 6. Drop unnecessary columns
# ============================================

# Example: Remove Year if it is not required
df = df.drop("Year", axis=1)

print("\nAfter dropping Year:")
print(df.head())


# ============================================
# 7. One-Hot Encoding
# ============================================

df = pd.get_dummies(df, columns=["Area"])

print("\nAfter One-Hot Encoding:")
print(df.head())


# ============================================
# 8. Sort dataset
# ============================================

df = df.sort_values(
    "average_rain_fall_mm_per_year",
    ascending=False
)

print("\nSorted by rainfall (highest first):")
print(df.head())
```
