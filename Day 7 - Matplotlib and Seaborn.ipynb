Introduction to Matplotlib
1. Theory
Matplotlib is a popular Python library for data visualization.
It is widely used for plotting graphs and charts, allowing developers to represent data visually.

Key features:

Produces publication-quality plots.
Supports line plots, bar charts, histograms, scatter plots, pie charts, etc.
Can integrate with NumPy and Pandas.
Provides control over figure size, labels, legends, colors, and styles.


Applications
Visualizing statistical data.
Understanding trends and patterns in datasets.
Exploratory Data Analysis (EDA).
Plotting real-time sensor data.
Creating dashboards and reports.
Used in scientific research, finance, engineering, and machine learning.

List of Functions and Their Usage
Function	Usage
plt.plot(x,y)	Line plot
plt.bar(x,y)	Bar chart
plt.hist(data, bins)	Histogram
plt.scatter(x,y)	Scatter plot
plt.pie(values, labels)	Pie chart
plt.xlabel(), plt.ylabel()	Set axis labels
plt.title()	Set plot title
plt.legend()	Add legend
plt.grid()	Show grid
plt.subplot()	Multiple plots in one figure
plt.figure(figsize=(w,h))	Define figure size
plt.show()	Display plot

Import and Execution

```python
# Importing Matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
x = np.linspace(0, 10, 10)
print (x)
y = np.sin(x)
print(y)

# Basic Line Plot
plt.plot(x, y, label="sin(x)", color="blue", linestyle="--")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Sine Wave Example")
plt.legend()
plt.grid(True)
plt.show()
```

Examples

```python
# Example 1: Line Plot with Multiple Lines
import numpy as np

y2 = np.cos(x)
print(y2)
plt.plot(x, y, label="sin(x)", color="red")
plt.plot(x, y2, label="cos(x)", color="green")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Sine and Cosine Waves")
plt.legend()
plt.show()

# Example 2: Bar Chart
categories = ['A','B','C','D']
values = [10, 24, 36, 18]
plt.bar(categories, values, color=['blue','orange','green','red'], label="categories")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Chart Example")
plt.legend ()
plt.show()

# Example 3: Histogram
#students = ['A','B','C','D']
data = np.random.randn(100)
print(data)
plt.hist(data, bins=100, color='purple', edgecolor='black',label="random")
plt.title("Histogram Example")
plt.legend()
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

#Example 3.1 Histogram (Students Marks)

marks = [10, 15, 25, 30, 35, 40, 45, 50, 52, 55, 60, 42, 48, 33, 28, 22, 18]

# Define bins (ranges)
bins = [0, 20, 40, 60]

# Create histogram
plt.hist(marks, bins=bins, edgecolor='black')

# Labels and title
plt.xlabel('Marks Range')
plt.ylabel('Number of Students')
plt.title('Histogram of Student Marks')

# Show plot
plt.show()

# Example 4: Scatter Plot
categories = ['A','B','C','D','E','F','G','H','I']
values = [10, 24, 36, 18,19,20,23,26,12]
plt.scatter(categories,values, color='darkgreen',label="categories")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Scatter Plot Example")
plt.legend()
plt.show()

# Example 5: Pie Chart
sizes = [60, 20, 25, 25, 50, 45, 23]
labels = ['Apple', 'Banana', 'Mango', 'Grapes', 'A', 'B', 'C']
plt.pie(sizes, labels=labels, autopct='%1.5f%%', startangle=90)
plt.title("Pie Chart Example")
plt.show()

# Example 6: Subplots
fig, axs = plt.subplots(2, 2, figsize=(8,8))

# Line Plot
axs[0,0].plot(x, y, color='blue')
axs[0,0].set_title("Line Plot")

# Bar Chart
axs[0,1].bar(categories, values, color='orange')
axs[0,1].set_title("Bar Chart")

# Histogram
axs[1,0].hist(marks, bins=bins, color='green')
axs[1,0].set_title("Histogram")

# Scatter Plot
axs[1,1].scatter(categories,values, color='red')
axs[1,1].set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```

1. Definition
Seaborn is a Python data visualization library based on Matplotlib.
It provides a high-level interface for drawing attractive and informative statistical graphics with fewer lines of code.



2. Applications
Creating visually appealing statistical plots.
Quick data exploration and analysis.
Works seamlessly with Pandas DataFrames.
Advanced color palettes and themes for better aesthetics.
Simplifies complex visualizations like heatmaps, categorical plots, regression plots, etc.


3. List of Commonly Used Functions in Seaborn
sns.scatterplot() – Creates scatter plots.
sns.lineplot() – Creates line plots.
sns.barplot() – Shows average values with confidence intervals.
sns.countplot() – Counts frequency of categorical data.
sns.boxplot() – Creates box-and-whisker plots.
sns.violinplot() – Combines boxplot and KDE for distribution.
sns.heatmap() – Creates heatmaps for correlation matrices or tabular data.
sns.pairplot() – Pairwise relationships in dataset.
sns.distplot() / sns.histplot() – Distribution of variables.
sns.regplot() – Regression line with scatter plot.
sns.catplot() – Plotting categorical data.

4. Implementation and Examples

```python
# Seaborn Implementation - All in One Code

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
tips = sns.load_dataset("tips")

# 1. Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex", style="time")
plt.title("Scatter Plot - Total Bill vs Tip")
plt.show()

# 2. Line Plot
plt.figure(figsize=(6,4))
sns.lineplot(x="size", y="total_bill", data=tips, ci=None)
plt.title("Line Plot - Total Bill vs Size")
plt.show()

# 3. Bar Plot
plt.figure(figsize=(6,4))
sns.barplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title("Bar Plot - Average Total Bill per Day")
plt.show()

# 4. Count Plot
plt.figure(figsize=(6,4))
sns.countplot(x="day", data=tips, hue="sex")
plt.title("Count Plot - Customers per Day")
plt.show()

# 5. Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title("Box Plot - Total Bill Distribution")
plt.show()

# 6. Violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(x="day", y="total_bill", data=tips, hue="sex", split=True)
plt.title("Violin Plot - Distribution by Day and Gender")
plt.show()

# 7. Heatmap (Correlation Matrix)
plt.figure(figsize=(6,4))
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()

# 8. Pairplot
sns.pairplot(tips, hue="sex")
plt.suptitle("Pairplot of Tips Dataset")
plt.show()

# 9. Distribution Plot (Histogram + KDE)
plt.figure(figsize=(6,4))
sns.histplot(tips["total_bill"], kde=True, bins=20)
plt.title("Distribution of Total Bill")
plt.show()

# 10. Regression Plot
plt.figure(figsize=(6,4))
sns.regplot(x="total_bill", y="tip", data=tips)
plt.title("Regression Plot - Tip vs Total Bill")
plt.show()
```
